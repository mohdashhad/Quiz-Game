```python
import tkinter as tk
from tkinter import messagebox
import pygame
import pyttsx3
import random
import sys
import time
import math

# Initialize pygame for audio and display
pygame.mixer.init()
pygame.init()

# Initialize text-to-speech
engine = pyttsx3.init()
engine.setProperty('rate', 180)  # Slower, dramatic speech rate
engine.setProperty('voice', 'english')  # Default voice

# Load KBC audio files
try:
    intro_sound = pygame.mixer.Sound("kbc_intromusic.mp3.mpeg")
    suspense_sound = pygame.mixer.Sound("kbc_suspense.mp3.mpeg")
    correct_sound = pygame.mixer.Sound("kbc_correct.mp3.mpeg")
    wrong_sound = pygame.mixer.Sound("kbc_wrong.mp3.mpeg")
    tick_sound = pygame.mixer.Sound("kbc_timer.mp3")
    fireworks_sound = pygame.mixer.Sound("fireworks-29629.mp3")  # Fireworks sound
except pygame.error as e:
    print(f"Audio loading error: {e}")
    intro_sound = suspense_sound = correct_sound = wrong_sound = tick_sound = fireworks_sound = None

# Question database
questions = [
    # Basic Level (₹1,000 - ₹10,000)
    {"question": "How many days are there in a week?", "options": ["A) 5", "B) 6", "C) 7", "D) 8"], "answer": "C", "prize": "₹1,000"},
    {"question": "What is 2 + 2?", "options": ["A) 3", "B) 4", "C) 5", "D) 6"], "answer": "B", "prize": "₹1,000"},
    {"question": "Which color is the sky on a clear day?", "options": ["A) Green", "B) Blue", "C) Red", "D) Yellow"], "answer": "B", "prize": "₹1,000"},
    {"question": "How many sides does a square have?", "options": ["A) 3", "B) 4", "C) 5", "D) 6"], "answer": "B", "prize": "₹2,000"},
    {"question": "What is the opposite of hot?", "options": ["A) Warm", "B) Cold", "C) Cool", "D) Mild"], "answer": "B", "prize": "₹2,000"},
    {"question": "How many legs does a spider have?", "options": ["A) 6", "B) 8", "C) 10", "D) 12"], "answer": "B", "prize": "₹2,000"},
    {"question": "What is 5 + 3?", "options": ["A) 6", "B) 7", "C) 8", "D) 9"], "answer": "C", "prize": "₹3,000"},
    {"question": "Which animal is known as man’s best friend?", "options": ["A) Cat", "B) Dog", "C) Horse", "D) Bird"], "answer": "B", "prize": "₹3,000"},
    {"question": "How many colors are in a rainbow?", "options": ["A) 5", "B) 6", "C) 7", "D) 8"], "answer": "C", "prize": "₹3,000"},
    {"question": "What is the shape of a ball?", "options": ["A) Square", "B) Circle", "C) Triangle", "D) Sphere"], "answer": "D", "prize": "₹5,000"},
    {"question": "Which day comes after Monday?", "options": ["A) Sunday", "B) Tuesday", "C) Wednesday", "D) Thursday"], "answer": "B", "prize": "₹5,000"},
    {"question": "How many fingers are on one hand?", "options": ["A) 3", "B) 4", "C) 5", "D) 6"], "answer": "C", "prize": "₹5,000"},
    {"question": "What is the capital of India?", "options": ["A) Mumbai", "B) Delhi", "C) Kolkata", "D) Chennai"], "answer": "B", "prize": "₹10,000"},
    {"question": "How many months are in a year?", "options": ["A) 10", "B) 11", "C) 12", "D) 13"], "answer": "C", "prize": "₹10,000"},
    {"question": "Which fruit is yellow and curved?", "options": ["A) Apple", "B) Banana", "C) Orange", "D) Grape"], "answer": "B", "prize": "₹10,000"},
    {"question": "What is 10 - 4?", "options": ["A) 5", "B) 6", "C) 7", "D) 8"], "answer": "B", "prize": "₹1,000"},
    {"question": "Which season comes after summer?", "options": ["A) Winter", "B) Spring", "C) Autumn", "D) Monsoon"], "answer": "C", "prize": "₹2,000"},
    {"question": "How many wheels does a bicycle have?", "options": ["A) 1", "B) 2", "C) 3", "D) 4"], "answer": "B", "prize": "₹3,000"},
    {"question": "What do you use to write on a blackboard?", "options": ["A) Pen", "B) Pencil", "C) Chalk", "D) Marker"], "answer": "C", "prize": "₹5,000"},
    {"question": "Which animal is the king of the jungle?", "options": ["A) Tiger", "B) Lion", "C) Elephant", "D) Bear"], "answer": "B", "prize": "₹10,000"},
    {"question": "What is 3 × 3?", "options": ["A) 6", "B) 9", "C) 12", "D) 15"], "answer": "B", "prize": "₹1,000"},
    {"question": "Which bird lays the largest eggs?", "options": ["A) Sparrow", "B) Ostrich", "C) Eagle", "D) Penguin"], "answer": "B", "prize": "₹2,000"},
    {"question": "What is the smell of rain like?", "options": ["A) Sweet", "B) Bitter", "C) Wet", "D) Sour"], "answer": "C", "prize": "₹3,000"},
    {"question": "How many players are in a cricket team?", "options": ["A) 9", "B) 10", "C) 11", "D) 12"], "answer": "C", "prize": "₹5,000"},
    {"question": "What is the main ingredient in bread?", "options": ["A) Sugar", "B) Flour", "C) Salt", "D) Water"], "answer": "B", "prize": "₹10,000"},
    {"question": "Which direction does the sun rise from?", "options": ["A) West", "B) East", "C) North", "D) South"], "answer": "B", "prize": "₹1,000"},
    {"question": "How many hours are in a day?", "options": ["A) 12", "B) 24", "C) 36", "D) 48"], "answer": "B", "prize": "₹2,000"},
    {"question": "What is the boiling point of water in Celsius?", "options": ["A) 50", "B) 75", "C) 100", "D) 125"], "answer": "C", "prize": "₹3,000"},
    {"question": "Which is the smallest planet in our solar system?", "options": ["A) Earth", "B) Mars", "C) Mercury", "D) Venus"], "answer": "C", "prize": "₹5,000"},
    {"question": "What do bees produce?", "options": ["A) Milk", "B) Honey", "C) Butter", "D) Cheese"], "answer": "B", "prize": "₹10,000"},
    # Mid-Level (₹20,000 - ₹6,40,000)
    {"question": "Which planet is known as the Red Planet?", "options": ["A) Venus", "B) Mars", "C) Jupiter", "D) Saturn"], "answer": "B", "prize": "₹20,000"},
    {"question": "How many continents are there on Earth?", "options": ["A) 5", "B) 6", "C) 7", "D) 8"], "answer": "C", "prize": "₹20,000"},
    {"question": "Who was the first Prime Minister of India?", "options": ["A) Mahatma Gandhi", "B) Jawaharlal Nehru", "C) Sardar Patel", "D) Indira Gandhi"], "answer": "B", "prize": "₹40,000"},
    {"question": "Which gas do plants absorb from the atmosphere?", "options": ["A) Oxygen", "B) Nitrogen", "C) Carbon Dioxide", "D) Hydrogen"], "answer": "C", "prize": "₹40,000"},
    {"question": "In which year did India gain independence?", "options": ["A) 1945", "B) 1947", "C) 1950", "D) 1960"], "answer": "B", "prize": "₹80,000"},
    {"question": "What is the currency of Japan?", "options": ["A) Yuan", "B) Yen", "C) Won", "D) Ringgit"], "answer": "B", "prize": "₹80,000"},
    {"question": "Which river is the longest in the world?", "options": ["A) Amazon", "B) Nile", "C) Yangtze", "D) Mississippi"], "answer": "B", "prize": "₹1,60,000"},
    {"question": "Who painted the Mona Lisa?", "options": ["A) Vincent van Gogh", "B) Leonardo da Vinci", "C) Pablo Picasso", "D) Michelangelo"], "answer": "B", "prize": "₹1,60,000"},
    {"question": "What is the largest organ in the human body?", "options": ["A) Heart", "B) Liver", "C) Skin", "D) Brain"], "answer": "C", "prize": "₹3,20,000"},
    {"question": "Which country hosted the 2016 Summer Olympics?", "options": ["A) China", "B) Brazil", "C) Japan", "D) UK"], "answer": "B", "prize": "₹3,20,000"},
    {"question": "What is the chemical symbol for gold?", "options": ["A) Au", "B) Ag", "C) Fe", "D) Cu"], "answer": "A", "prize": "₹6,40,000"},
    {"question": "Which is the largest desert in the world?", "options": ["A) Sahara", "B) Gobi", "C) Antarctic", "D) Arabian"], "answer": "C", "prize": "₹20,000"},
    {"question": "Who wrote the Indian National Anthem?", "options": ["A) Mahatma Gandhi", "B) Rabindranath Tagore", "C) Bankim Chandra", "D) Subhas Bose"], "answer": "B", "prize": "₹40,000"},
    {"question": "Which ocean is the largest?", "options": ["A) Atlantic", "B) Indian", "C) Arctic", "D) Pacific"], "answer": "D", "prize": "₹80,000"},
    {"question": "What gas makes up most of Earth’s atmosphere?", "options": ["A) Oxygen", "B) Nitrogen", "C) Carbon Dioxide", "D) Helium"], "answer": "B", "prize": "₹1,60,000"},
    {"question": "Which country is known as the Land of the Rising Sun?", "options": ["A) China", "B) Japan", "C) Korea", "D) Thailand"], "answer": "B", "prize": "₹3,20,000"},
    {"question": "What is the capital of Brazil?", "options": ["A) Rio de Janeiro", "B) São Paulo", "C) Brasília", "D) Salvador"], "answer": "C", "prize": "₹6,40,000"},
    {"question": "Who invented the telephone?", "options": ["A) Thomas Edison", "B) Alexander Graham Bell", "C) Nikola Tesla", "D) Guglielmo Marconi"], "answer": "B", "prize": "₹20,000"},
    {"question": "Which animal is the tallest on Earth?", "options": ["A) Elephant", "B) Giraffe", "C) Blue Whale", "D) Ostrich"], "answer": "B", "prize": "₹40,000"},
    {"question": "What is the smallest country by land area?", "options": ["A) Monaco", "B) Vatican City", "C) San Marino", "D) Liechtenstein"], "answer": "B", "prize": "₹80,000"},
    {"question": "Which element is the most abundant in the universe?", "options": ["A) Oxygen", "B) Hydrogen", "C) Carbon", "D) Helium"], "answer": "B", "prize": "₹1,60,000"},
    {"question": "Who directed the movie 'Titanic'?", "options": ["A) Steven Spielberg", "B) James Cameron", "C) Christopher Nolan", "D) Quentin Tarantino"], "answer": "B", "prize": "₹3,20,000"},
    {"question": "What is the capital of Australia?", "options": ["A) Sydney", "B) Melbourne", "C) Canberra", "D) Perth"], "answer": "C", "prize": "₹6,40,000"},
    {"question": "Which Indian festival is known as the festival of lights?", "options": ["A) Holi", "B) Diwali", "C) Dussehra", "D) Raksha Bandhan"], "answer": "B", "prize": "₹20,000"},
    {"question": "What is the freezing point of water in Celsius?", "options": ["A) 0", "B) 32", "C) 100", "D) -10"], "answer": "A", "prize": "₹40,000"},
    {"question": "Which mountain range separates Europe and Asia?", "options": ["A) Alps", "B) Himalayas", "C) Ural", "D) Andes"], "answer": "C", "prize": "₹80,000"},
    {"question": "Who wrote 'Romeo and Juliet'?", "options": ["A) Charles Dickens", "B) William Shakespeare", "C) Jane Austen", "D) Mark Twain"], "answer": "B", "prize": "₹1,60,000"},
    {"question": "Which country has the most population?", "options": ["A) India", "B) China", "C) USA", "D) Russia"], "answer": "B", "prize": "₹3,20,000"},
    {"question": "What is the main source of energy for Earth?", "options": ["A) Moon", "B) Sun", "C) Wind", "D) Oceans"], "answer": "B", "prize": "₹6,40,000"},
    {"question": "Which Indian city is known as the Pink City?", "options": ["A) Jaipur", "B) Jodhpur", "C) Udaipur", "D) Agra"], "answer": "A", "prize": "₹20,000"},
    {"question": "What is the chemical formula for water?", "options": ["A) H2O", "B) CO2", "C) O2", "D) N2"], "answer": "A", "prize": "₹40,000"},
    {"question": "Which planet has the most moons?", "options": ["A) Jupiter", "B) Saturn", "C) Uranus", "D) Neptune"], "answer": "B", "prize": "₹80,000"},
    {"question": "Who was the first woman Prime Minister of India?", "options": ["A) Sonia Gandhi", "B) Indira Gandhi", "C) Pratibha Patil", "D) Sushma Swaraj"], "answer": "B", "prize": "₹1,60,000"},
    {"question": "What is the largest mammal in the world?", "options": ["A) Elephant", "B) Blue Whale", "C) Giraffe", "D) Hippopotamus"], "answer": "B", "prize": "₹3,20,000"},
    {"question": "Which gas is used in balloons to make them float?", "options": ["A) Oxygen", "B) Nitrogen", "C) Helium", "D) Carbon Dioxide"], "answer": "C", "prize": "₹6,40,000"},
    {"question": "Which Indian state is known as the 'Land of Five Rivers'?", "options": ["A) Punjab", "B) Haryana", "C) Gujarat", "D) Rajasthan"], "answer": "A", "prize": "₹20,000"},
    {"question": "What is the capital of France?", "options": ["A) London", "B) Paris", "C) Berlin", "D) Rome"], "answer": "B", "prize": "₹40,000"},
    {"question": "Which instrument is used to measure temperature?", "options": ["A) Barometer", "B) Thermometer", "C) Anemometer", "D) Hydrometer"], "answer": "B", "prize": "₹80,000"},
    {"question": "Who is known as the Father of the Nation in India?", "options": ["A) Jawaharlal Nehru", "B) Mahatma Gandhi", "C) Bhagat Singh", "D) Subhas Bose"], "answer": "B", "prize": "₹1,60,000"},
    {"question": "Which country is home to the kangaroo?", "options": ["A) Australia", "B) New Zealand", "C) South Africa", "D) India"], "answer": "A", "prize": "₹3,20,000"},
    # Tough Level (₹12,50,000 - ₹7,00,00,000)
    {"question": "Who discovered the theory of relativity?", "options": ["A) Isaac Newton", "B) Albert Einstein", "C) Galileo Galilei", "D) Nikola Tesla"], "answer": "B", "prize": "₹12,50,000"},
    {"question": "Which Indian state has the longest coastline?", "options": ["A) Gujarat", "B) Maharashtra", "C) Tamil Nadu", "D) Andhra Pradesh"], "answer": "A", "prize": "₹12,50,000"},
    {"question": "In which year was the first Indian satellite, Aryabhata, launched?", "options": ["A) 1975", "B) 1980", "C) 1969", "D) 1985"], "answer": "A", "prize": "₹25,00,000"},
    {"question": "Which Indian mathematician is known for the concept of zero?", "options": ["A) Aryabhata", "B) Brahmagupta", "C) Bhaskara", "D) Ramanujan"], "answer": "B", "prize": "₹25,00,000"},
    {"question": "Which element has the highest melting point?", "options": ["A) Iron", "B) Tungsten", "C) Gold", "D) Platinum"], "answer": "B", "prize": "₹50,00,000"},
    {"question": "Who was the first Indian to win a Nobel Prize?", "options": ["A) C.V. Raman", "B) Rabindranath Tagore", "C) Amartya Sen", "D) Mother Teresa"], "answer": "B", "prize": "₹50,00,000"},
    {"question": "Which is the deepest point in the Earth’s oceans?", "options": ["A) Mariana Trench", "B) Tonga Trench", "C) Puerto Rico Trench", "D) Java Trench"], "answer": "A", "prize": "₹1,00,00,000"},
    {"question": "Who composed the music for the Indian National Song 'Vande Mataram'?", "options": ["A) A.R. Rahman", "B) Bankim Chandra", "C) Rabindranath Tagore", "D) Lata Mangeshkar"], "answer": "C", "prize": "₹1,00,00,000"},
    {"question": "Which Indian emperor built the Taj Mahal?", "options": ["A) Akbar", "B) Shah Jahan", "C) Aurangzeb", "D) Babur"], "answer": "B", "prize": "₹3,00,00,000"},
    {"question": "What is the rarest blood type in humans?", "options": ["A) O+", "B) AB-", "C) B-", "D) A+"], "answer": "B", "prize": "₹3,00,00,000"},
    {"question": "Which scientist discovered penicillin?", "options": ["A) Louis Pasteur", "B) Alexander Fleming", "C) Jonas Salk", "D) Robert Koch"], "answer": "B", "prize": "₹7,00,00,000"},
    {"question": "In which year did the Titanic sink?", "options": ["A) 1910", "B) 1912", "C) 1914", "D) 1916"], "answer": "B", "prize": "₹7,00,00,000"},
    {"question": "Which country was the first to grant women the right to vote?", "options": ["A) USA", "B) UK", "C) New Zealand", "D) France"], "answer": "C", "prize": "₹12,50,000"},
    {"question": "What is the longest bone in the human body?", "options": ["A) Spine", "B) Femur", "C) Tibia", "D) Humerus"], "answer": "B", "prize": "₹25,00,000"},
    {"question": "Which gas, discovered on the sun before Earth, is the second most abundant element in the universe?", "options": ["A) Helium", "B) Hydrogen", "C) Oxygen", "D) Nitrogen"], "answer": "A", "prize": "₹50,00,000"},
    {"question": "Who was the first person to fly solo across the Atlantic Ocean?", "options": ["A) Amelia Earhart", "B) Charles Lindbergh", "C) Wilbur Wright", "D) Neil Armstrong"], "answer": "B", "prize": "₹1,00,00,000"},
    {"question": "Which Indian state was the first to implement the Right to Education Act?", "options": ["A) Kerala", "B) Tamil Nadu", "C) Sikkim", "D) Maharashtra"], "answer": "C", "prize": "₹3,00,00,000"},
    {"question": "What is the primary source of energy for the Earth’s climate system?", "options": ["A) Geothermal", "B) Solar", "C) Tidal", "D) Wind"], "answer": "B", "prize": "₹7,00,00,000"},
    {"question": "Which element is named after the Greek word for 'hidden'?", "options": ["A) Krypton", "B) Xenon", "C) Argon", "D) Neon"], "answer": "A", "prize": "₹12,50,000"},
    {"question": "Which Indian scientist won the Nobel Prize for Physics in 1930?", "options": ["A) Homi Bhabha", "B) C.V. Raman", "C) Satyendra Bose", "D) Meghnad Saha"], "answer": "B", "prize": "₹25,00,000"},
    {"question": "What is the smallest bone in the human body?", "options": ["A) Stapes", "B) Incus", "C) Malleus", "D) Phalanx"], "answer": "A", "prize": "₹50,00,000"},
    {"question": "Which country has the most deserts in the world?", "options": ["A) Australia", "B) Antarctica", "C) Saudi Arabia", "D) India"], "answer": "B", "prize": "₹1,00,00,000"},
    {"question": "Which Indian freedom fighter wrote the book 'Gita Rahasya'?", "options": ["A) Bal Gangadhar Tilak", "B) Lala Lajpat Rai", "C) Bipin Chandra Pal", "D) Gopal Krishna Gokhale"], "answer": "A", "prize": "₹3,00,00,000"},
    {"question": "Which is the only mammal capable of true flight?", "options": ["A) Bat", "B) Flying Squirrel", "C) Pterosaur", "D) Bird"], "answer": "A", "prize": "₹7,00,00,000"},
    {"question": "Which river flows through the Grand Canyon?", "options": ["A) Mississippi", "B) Colorado", "C) Missouri", "D) Rio Grande"], "answer": "B", "prize": "₹12,50,000"},
    {"question": "Which Indian city hosted the first Asian Games?", "options": ["A) Mumbai", "B) Kolkata", "C) New Delhi", "D) Chennai"], "answer": "C", "prize": "₹25,00,000"},
    {"question": "What is the hardest naturally occurring substance known?", "options": ["A) Diamond", "B) Quartz", "C) Graphite", "D) Topaz"], "answer": "A", "prize": "₹50,00,000"},
    {"question": "Which mathematician developed the Fibonacci sequence?", "options": ["A) Euclid", "B) Fibonacci", "C) Pythagoras", "D) Archimedes"], "answer": "B", "prize": "₹1,00,00,000"},
    {"question": "Which planet has the shortest day?", "options": ["A) Jupiter", "B) Saturn", "C) Earth", "D) Mars"], "answer": "A", "prize": "₹3,00,00,000"},
    {"question": "Which Indian state was formed last?", "options": ["A) Telangana", "B) Jharkhand", "C) Chhattisgarh", "D) Uttarakhand"], "answer": "A", "prize": "₹7,00,00,000"},
]

# Prize ladder
prize_ladder = [
    "₹7,00,00,000", "₹3,00,00,000", "₹1,00,00,000", "₹50,00,000", "₹25,00,000",
    "₹12,50,000", "₹6,40,000", "₹3,20,000", "₹1,60,000", "₹80,000",
    "₹40,000", "₹20,000", "₹10,000", "₹5,000", "₹3,000", "₹2,000", "₹1,000"
]

# Splash Screen Function
def show_splash_screen():
    SCREEN_WIDTH, SCREEN_HEIGHT = 800, 600
    splash_screen = pygame.display.set_mode((SCREEN_WIDTH, SCREEN_HEIGHT))
    pygame.display.set_caption("KBC Splash")

    # Load and scale the KBC logo
    logo = pygame.image.load("KBC_logo.jpeg")
    logo = pygame.transform.scale(logo, (SCREEN_WIDTH, SCREEN_HEIGHT))

    # Display logo
    splash_screen.blit(logo, (0, 0))
    pygame.display.flip()

    # Play intro sound
    if intro_sound:
        intro_sound.play()
        pygame.time.wait(int(intro_sound.get_length() * 1000))  # Dynamic wait based on audio length

    pygame.display.quit()  # Close display

# Game class
class KBCGame:
    def __init__(self, root):
        self.root = root
        self.root.title("Kaun Banega Crorepati")
        self.root.attributes("-fullscreen", True)
        self.root.configure(bg="#0A0A23")
        self.root.withdraw()  # Hide the window initially

        self.money = "₹0"
        self.current_question = 0
        self.lifelines = {"50:50": True, "Audience": True, "Phone": True}
        self.is_paused = False
        self.game_questions = self.select_questions()
        self.time_left = 30
        self.answer_selected = False
        self.timer_running = False
        self.tick_sound = tick_sound  # Store tick sound for timer
        self.max_border_width = 10  # Starting border width
        self.min_border_width = 2   # Minimum border width
        self.fireworks = []
        self.fireworks_duration = 34000  # Duration of fireworks in ms (34 seconds)
        self.fireworks_active = False
        self.congrats_label = None  # Initialize congrats label

        # Show splash screen before GUI
        show_splash_screen()

        # Show the GUI after splash
        self.root.deiconify()  # Show the window
        self.create_widgets()
        self.next_question()

    def select_questions(self):
        selected = []
        for prize in reversed(prize_ladder):
            level_questions = [q for q in questions if q["prize"] == prize]
            if level_questions:
                selected.append(random.choice(level_questions))
        return selected

    def bind_hover(self):
        """Bind hover effect to option buttons."""
        for btn in self.option_buttons:
            btn.bind("<Enter>", lambda e, b=btn: b.config(bg="#FFD700"))
            btn.bind("<Leave>", lambda e, b=btn: b.config(bg="#1F1F5A"))
            btn.config(bg="#1F1F5A")  # Ensure default color

    def unbind_hover(self):
        """Unbind hover effect from option buttons."""
        for btn in self.option_buttons:
            btn.unbind("<Enter>")
            btn.unbind("<Leave>")
            btn.config(bg="#1F1F5A")  # Reset color

    def create_widgets(self):
        # Title
        self.title = tk.Label(self.root, text="Kaun Banega Crorepati", font=("Arial", 36, "bold"), fg="gold", bg="#0A0A23")
        self.title.pack(pady=20)

        # Control buttons frame
        self.control_frame = tk.Frame(self.root, bg="#0A0A23")
        self.control_frame.pack(side=tk.TOP, anchor=tk.NE, padx=20, pady=10)
        tk.Button(self.control_frame, text="Play", font=("Arial", 12), fg="white", bg="green", command=self.play_game).pack(side=tk.LEFT, padx=5)
        tk.Button(self.control_frame, text="Pause", font=("Arial", 12), fg="white", bg="orange", command=self.pause_game).pack(side=tk.LEFT, padx=5)
        tk.Button(self.control_frame, text="Quit", font=("Arial", 12), fg="white", bg="red", command=self.quit_game).pack(side=tk.LEFT, padx=5)

        # Prize ladder frame
        self.ladder_frame = tk.Frame(self.root, bg="#0A0A23")
        self.ladder_frame.pack(side=tk.RIGHT, padx=20)
        tk.Label(self.ladder_frame, text="Prize Money", font=("Arial", 18, "bold"), fg="white", bg="#0A0A23").pack()
        self.ladder_labels = []
        for prize in prize_ladder:
            prize_label = tk.Label(self.ladder_frame, text=prize, font=("Arial", 14), fg="white", bg="#0A0A23")
            prize_label.pack(pady=5)
            self.ladder_labels.append(prize_label)

        # Question frame
        self.question_frame = tk.Frame(self.root, bg="#0A0A23")
        self.question_frame.pack(pady=50)
        self.question_label = tk.Label(self.question_frame, text="", font=("Arial", 20, "bold"), fg="white", bg="#0A0A23", wraplength=800)
        self.question_label.pack()
        self.timer_canvas = tk.Canvas(self.question_frame, width=120, height=120, bg="#0A0A23", highlightthickness=0)
        self.timer_canvas.pack(pady=10)

        # Fireworks canvas (covers entire window)
        self.fireworks_canvas = tk.Canvas(self.root, bg="#0A0A23", highlightthickness=0)

        # Options frame
        self.options_frame = tk.Frame(self.root, bg="#0A0A23")
        self.options_frame.pack(pady=20)
        self.option_buttons = []
        for i in range(4):
            btn = tk.Button(self.options_frame, text="", font=("Arial", 16), fg="white", bg="#1F1F5A", width=30, command=lambda x=i: self.check_answer(x))
            btn.grid(row=i//2, column=i%2, padx=10, pady=10)
            self.option_buttons.append(btn)

        # Lifeline frame
        self.lifeline_frame = tk.Frame(self.root, bg="#0A0A23")
        self.lifeline_frame.pack(pady=20)
        tk.Button(self.lifeline_frame, text="50:50", font=("Arial", 14), fg="white", bg="#1F1F5A", command=self.fifty_fifty).pack(side=tk.LEFT, padx=10)
        tk.Button(self.lifeline_frame, text="Audience Poll", font=("Arial", 14), fg="white", bg="#1F1F5A", command=self.audience_poll).pack(side=tk.LEFT, padx=10)
        tk.Button(self.lifeline_frame, text="Phone a Friend", font=("Arial", 14), fg="white", bg="#1F1F5A", command=self.phone_friend).pack(side=tk.LEFT, padx=10)

    def play_game(self):
        if self.is_paused:
            if suspense_sound:
                suspense_sound.play()
            self.is_paused = False
            if self.timer_running:
                self.countdown()

    def pause_game(self):
        if not self.is_paused:
            if suspense_sound:
                suspense_sound.stop()
            if self.tick_sound:
                self.tick_sound.stop()  # Stop tick sound when pausing
            self.is_paused = True

    def start_timer(self):
        self.time_left = 30
        self.answer_selected = False
        self.timer_running = True
        self.timer_canvas.delete("all")
        self.timer_canvas.create_oval(10, 10, 110, 110, outline='#ffd700', width=self.max_border_width, fill='#0A0A23')
        self.countdown()

    def countdown(self):
        if self.time_left > 0 and not self.answer_selected and not self.is_paused and self.timer_running:
            self.timer_canvas.delete("all")
            angle = (self.time_left / 30) * 360
            # Calculate border width based on remaining time
            progress = self.time_left / 30  # 1.0 at start, 0.0 at end
            current_border_width = self.min_border_width + (self.max_border_width - self.min_border_width) * progress
            # Change outline color based on time left
            outline_color = '#ffd700'  # Gold
            if self.time_left <= 10:
                outline_color = '#ff0000'  # Red
            elif self.time_left <= 20:
                outline_color = '#ffa500'  # Orange
            self.timer_canvas.create_arc(
                10, 10, 110, 110, start=0, extent=angle, 
                outline=outline_color, width=int(current_border_width), style='arc'
            )
            self.timer_canvas.create_text(
                60, 60, text=f"{self.time_left}", fill='#ffd700', 
                font=('Arial', 20, 'bold'), tags="timer"
            )
            if self.tick_sound:
                self.tick_sound.play()  # Play tick sound each second
            self.time_left -= 1
            self.root.after(1000, self.countdown)
        elif self.time_left == 0 and not self.answer_selected:
            self.handle_timeout()

    def handle_timeout(self):
        self.timer_running = False
        if suspense_sound:
            suspense_sound.stop()
        if self.tick_sound:
            self.tick_sound.stop()  # Stop tick sound on timeout
        if wrong_sound:
            wrong_sound.play()
        self.timer_canvas.delete("all")
        self.timer_canvas.create_text(60, 60, text="Time's Up!", fill='#ffd700', font=('Arial', 20, 'bold'))
        messagebox.showinfo("Time's Up!", f"Game Over! You leave with {self.money}")
        self.show_play_again()

    def next_question(self):
        if self.current_question < len(self.game_questions):
            q = self.game_questions[self.current_question]
            self.question_label.config(text=f"Question for {q['prize']}: {q['question']}")
            for i, btn in enumerate(self.option_buttons):
                btn.config(text=q["options"][i], state="normal", bg="#1F1F5A")
            for i, label in enumerate(self.ladder_labels):
                if label.cget("text") == q["prize"]:
                    label.config(fg="gold")
                else:
                    label.config(fg="white")
            self.bind_hover()  # Enable hover effect
            self.root.update()
            if suspense_sound:
                suspense_sound.play()
                pygame.time.wait(1000)
            engine.say(f"Question for {q['prize']}. {q['question']}")
            engine.runAndWait()  # Wait for speech to finish
            self.start_timer()  # Start timer after speech
        else:
            self.timer_running = False
            if self.tick_sound:
                self.tick_sound.stop()  # Stop tick sound at game end
            messagebox.showinfo("Congratulations!", f"You've won {self.game_questions[-1]['prize']}!")
            self.start_fireworks()

    def check_answer(self, idx):
        if self.is_paused or self.answer_selected:
            return
        self.answer_selected = True
        self.timer_running = False
        if suspense_sound:
            suspense_sound.stop()
        if self.tick_sound:
            self.tick_sound.stop()  # Stop tick sound when answer is selected
        self.timer_canvas.delete("all")  # Clear timer canvas after answer
        q = self.game_questions[self.current_question]
        selected = chr(65 + idx)
        if selected == q["answer"]:
            if correct_sound:
                correct_sound.play()
            self.money = q["prize"]
            messagebox.showinfo("Correct!", f"You've won {self.money}!")
            self.current_question += 1
            self.next_question()
        else:
            if wrong_sound:
                wrong_sound.play()
            messagebox.showinfo("Wrong!", f"Game Over! You leave with {self.money}")
            self.show_play_again()

    def fifty_fifty(self):
        if self.is_paused or not self.lifelines["50:50"]:
            return
        q = self.game_questions[self.current_question]
        correct_idx = ord(q["answer"]) - 65
        wrong_options = [i for i in range(4) if i != correct_idx]
        to_remove = random.sample(wrong_options, 2)
        for i in to_remove:
            self.option_buttons[i].config(text="", state="disabled", bg="#1F1F5A")
        self.lifelines["50:50"] = False

    def audience_poll(self):
        if self.is_paused or not self.lifelines["Audience"]:
            return
        q = self.game_questions[self.current_question]
        correct_idx = ord(q["answer"]) - 65
        poll = [random.randint(10, 30) for _ in range(4)]
        poll[correct_idx] = 100 - sum(poll) + poll[correct_idx]
        poll_text = f"A: {poll[0]}%  B: {poll[1]}%  C: {poll[2]}%  D: {poll[3]}%"
        print(f"Audience Poll called. Displaying: {poll_text}")  # Debug print
        messagebox.showinfo("Audience Poll", poll_text)
        self.lifelines["Audience"] = False

    def phone_friend(self):
        if self.is_paused or not self.lifelines["Phone"]:
            return
        q = self.game_questions[self.current_question]
        phone_text = f"Your friend suggests: {q['answer']}"
        print(f"Phone a Friend called. Displaying: {phone_text}")  # Debug print
        messagebox.showinfo("Phone a Friend", phone_text)
        self.lifelines["Phone"] = False

    def start_fireworks(self):
        self.unbind_hover()  # Disable hover effect
        # Hide game elements
        self.title.pack_forget()
        self.control_frame.pack_forget()
        self.ladder_frame.pack_forget()
        self.question_frame.pack_forget()
        self.options_frame.pack_forget()
        self.lifeline_frame.pack_forget()

        # Configure and show fireworks canvas
        self.fireworks_canvas.config(width=self.root.winfo_screenwidth(), height=self.root.winfo_screenheight())
        self.fireworks_canvas.pack(fill="both", expand=True)
        if fireworks_sound:
            fireworks_sound.play()

        # Add Congratulations message
        self.congrats_label = tk.Label(
            self.fireworks_canvas, 
            text="Congratulations, You Won the Game!",
            font=("Arial", 36, "bold"),
            fg="gold",
            bg="#0A0A23"
        )
        self.congrats_label.place(relx=0.5, rely=0.3, anchor="center")

        # Initialize fireworks
        self.fireworks_active = True
        self.fireworks = []

        # Schedule firework bursts every 2-3 seconds for 34 seconds
        for i in range(0, self.fireworks_duration, 2500):  # Every 2.5 seconds
            self.root.after(i, lambda: self.create_firework(
                random.randint(self.root.winfo_screenwidth() // 4, 3 * self.root.winfo_screenwidth() // 4),
                random.randint(self.root.winfo_screenheight() // 4, self.root.winfo_screenheight() // 2)
            ))

        # Add Play Again button
        self.play_again_button = tk.Button(
            self.fireworks_canvas, text="Play Again", font=("Arial", 20, "bold"),
            fg="gold", bg="#1F1F5A", command=self.play_again
        )
        self.play_again_button.place(relx=0.5, rely=0.8, anchor="center")
        self.play_again_button.bind("<Enter>", lambda e: self.play_again_button.config(bg="#FFD700"))
        self.play_again_button.bind("<Leave>", lambda e: self.play_again_button.config(bg="#1F1F5A"))

        # Start updating fireworks
        self.update_fireworks()

    def create_firework(self, x, y):
        if not self.fireworks_active:
            return
        colors = ["red", "yellow", "blue", "green", "purple", "orange"]
        for _ in range(30):  # Increased number of particles for denser effect
            angle = random.uniform(0, 2 * math.pi)
            speed = random.uniform(5, 15)
            vx = speed * math.cos(angle)
            vy = speed * math.sin(angle)
            color = random.choice(colors)
            particle = {
                "x": x,
                "y": y,
                "vx": vx,
                "vy": vy,
                "color": color,
                "life": 100,  # Increased life for longer visibility (100 frames ~ 2 seconds at 20ms)
                "id": self.fireworks_canvas.create_oval(x-3, y-3, x+3, y+3, fill=color)
            }
            self.fireworks.append(particle)

    def update_fireworks(self):
        if not self.fireworks_active:
            return
        for particle in self.fireworks[:]:
            particle["x"] += particle["vx"]
            particle["y"] += particle["vy"]
            particle["vy"] += 0.1  # Gravity effect
            particle["life"] -= 1

            # Update particle position
            self.fireworks_canvas.coords(particle["id"], 
                                        particle["x"]-3, particle["y"]-3, 
                                        particle["x"]+3, particle["y"]+3)

            if particle["life"] <= 0:
                self.fireworks_canvas.delete(particle["id"])
                self.fireworks.remove(particle)

        # Continue updating if fireworks are active or particles remain
        if self.fireworks or self.fireworks_active:
            self.root.after(20, self.update_fireworks)
        else:
            self.fireworks_active = False

        # Stop fireworks after duration
        self.root.after(self.fireworks_duration, self.stop_fireworks)

    def stop_fireworks(self):
        self.fireworks_active = False
        # Clean up any remaining particles
        for particle in self.fireworks[:]:
            self.fireworks_canvas.delete(particle["id"])
            self.fireworks.remove(particle)

    def show_play_again(self):
        self.unbind_hover()  # Disable hover effect
        # Hide game elements
        self.title.pack_forget()
        self.control_frame.pack_forget()
        self.ladder_frame.pack_forget()
        self.question_frame.pack_forget()
        self.options_frame.pack_forget()
        self.lifeline_frame.pack_forget()

        # Show play again button on main canvas
        self.play_again_button = tk.Button(
            self.root, text="Play Again", font=("Arial", 20, "bold"),
            fg="gold", bg="#1F1F5A", command=self.play_again
        )
        self.play_again_button.place(relx=0.5, rely=0.5, anchor="center")
        self.play_again_button.bind("<Enter>", lambda e: self.play_again_button.config(bg="#FFD700"))
        self.play_again_button.bind("<Leave>", lambda e: self.play_again_button.config(bg="#1F1F5A"))

    def play_again(self):
        # Stop all sounds, including fireworks
        pygame.mixer.stop()  # Stops all Sound objects (fireworks, tick, etc.)
        
        # Reset game state
        self.money = "₹0"
        self.current_question = 0
        self.lifelines = {"50:50": True, "Audience": True, "Phone": True}
        self.is_paused = False
        self.game_questions = self.select_questions()
        self.time_left = 30
        self.answer_selected = False
        self.timer_running = False
        self.fireworks = []
        self.fireworks_active = False

        # Clear fireworks canvas and play again button
        self.fireworks_canvas.delete("all")
        self.fireworks_canvas.pack_forget()
        if hasattr(self, 'play_again_button'):
            self.play_again_button.destroy()
        if self.congrats_label:
            self.congrats_label.destroy()
            self.congrats_label = None

        # Restore game elements
        self.title.pack(pady=20)
        self.control_frame.pack(side=tk.TOP, anchor=tk.NE, padx=20, pady=10)
        self.ladder_frame.pack(side=tk.RIGHT, padx=20)
        self.question_frame.pack(pady=50)
        self.options_frame.pack(pady=20)
        self.lifeline_frame.pack(pady=20)

        self.unbind_hover()  # Ensure hover is off before new question
        # Start new game
        self.next_question()

    def quit_game(self):
        self.unbind_hover()  # Disable hover effect
        self.timer_running = False
        if suspense_sound:
            suspense_sound.stop()
        if self.tick_sound:
            self.tick_sound.stop()  # Stop tick sound when quitting
        if self.congrats_label:
            self.congrats_label.destroy()
            self.congrats_label = None
        messagebox.showinfo("Quit", f"You walk away with {self.money}")
        self.root.destroy()  # Exit the game completely

# Run the game
if __name__ == "__main__":
    root = tk.Tk()
    game = KBCGame(root)
    root.mainloop()
#CREATED BY ASHHAD    
```
